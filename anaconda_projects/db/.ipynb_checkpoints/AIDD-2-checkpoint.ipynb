### 8. 判断结构

```python
score = 65
passed = 60
if score > passed:
    print("通过")
else:
    print("挂科")
```

### 9. 循环结构

```python
scores_dict = {"小明":62, "小红":82, "小黑":58}
passed = 60
for student in scores_dict:
    if scores_dict[student] > passed:
        print ("%s通过"%(student))
    else:
        print ("%s挂科"%(student))
```

### 14. 类 class （重要！面向对象）

```python
class people: 
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def introduce(self):
        print ("I'm %s, I'm %d years old!"%(self.name, self.age))
```

```python
a=people("xiaoming",15)
```

```python
a.introduce()
```

```python
class student(people): #people的子类，继承父类people的方法
    def __init__(self, name, age, score=0):
        super().__init__(name,age)
        self.set_score(score)
    def set_score(self,score):
        self.score = score
    def report_score(self):
        print ("My score is %.1f"%self.score)
```

```python
b = student("xiaohei",12)
```

```python
b.introduce()
```

```python
b.report_score()
```

```python
b.set_score(59)
```

```python
b.report_score()
```
